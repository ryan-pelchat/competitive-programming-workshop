```python
# https://stackoverflow.com/a/16926357 for zero padding
mask = 0b1011
subset = mask
ans = []
print(f"Solving proper subset for: {bin(mask)[2:]}")
while subset:
  print("-"*50)
  print(f"\t\t\t{format((subset), '#06b')[2:]} (subset)")
  print(f"\t\t\t{format((subset -1), '#06b')[2:]} (subset -1)")
  print("."*24)
  print(f"\t\t\t{format((mask), '#06b')[2:]} (mask)") ##010b
  print(f"\t\t\t{format((subset -1), '#06b')[2:]} (subset -1)")
  #print(f"\t {bin(subset - 1)[2:]} (subset -1)")
  print(" "*24+"-"*15)
  print(f"Element in Subset ->\t{format(mask & (subset - 1), '#06b')[2:]} (mask & (subset - 1)")
  subset = (mask & (subset - 1))
  # print(f"Element in Subset ->\t{format((subset), '#06b')[2:]}")
  ans.append(format((subset), '#06b')[2:])

print("\nAnswer:")
print("{" + str(ans)[1:-1] + "}")
# for result in ans:
#   print(f"ans: {result}")
```
